```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")

file_path = r"D:\Customer Segmentation Analysis\data\raw\customer_segmentation_dataset.xlsx"

# Loading datasets
customers = pd.read_excel(file_path, sheet_name="Customers")
transactions = pd.read_excel(file_path, sheet_name="Transactions")
products = pd.read_excel(file_path, sheet_name="Products")
annual_summary = pd.read_excel(file_path, sheet_name="Annual Summary")

print("All datasets loaded successfully!")
```

```python
# Missing values:

print("CUSTOMERS MISSING VALUES")
print(customers.isnull().sum())

print("\nTRANSACTIONS MISSING VALUES")
print(transactions.isnull().sum())

print("\nPRODUCTS MISSING VALUES")
print(products.isnull().sum())

print("\nANNUAL SUMMARY MISSING VALUES")
print(annual_summary.isnull().sum())
```

```python
# Duplicate records:

print("Customers Duplicates:", customers.duplicated().sum())

print("Transactions Duplicates:", transactions.duplicated().sum())

print("Products Duplicates:", products.duplicated().sum())

print("Annual Summary Duplicates:", annual_summary.duplicated().sum())
```

```python
# Checking Data types

print("CUSTOMERS DATA TYPES")
print(customers.dtypes)

print("\nTRANSACTIONS DATA TYPES")
print(transactions.dtypes)

print("\nPRODUCTS DATA TYPES")
print(products.dtypes)

print("\nANNUAL SUMMARY DATA TYPES")
print(annual_summary.dtypes)
```

```python
# converting date columns:

customers["SignupDate"] = pd.to_datetime(customers["SignupDate"])

transactions["OrderDate"] = pd.to_datetime(transactions["OrderDate"])

print("Date columns converted successfully!")
```

```python
# checking invalid values:

print("Invalid transaction quantities:",
      (transactions["Quantity"] <= 0).sum())
print("Invalid unit prices:",
      (transactions["UnitPrice"] < 0).sum())
print("Invalid sales:",
      (transactions["Sales"] < 0).sum())
print("Invalid profits:",
      (transactions["Profit"] < 0).sum())
print("Invalid discounts:",
      ((transactions["Discount"] < 0) | (transactions["Discount"] > 1)).sum())
```

```python
# Customers without matching customer records
print(
    "Transactions with unknown customers:",
    (~transactions["CustomerID"].isin(customers["CustomerID"])).sum()
)

# Transactions with unknown products
print(
    "Transactions with unknown products:",
    (~transactions["ProductID"].isin(products["ProductID"])).sum()
)
```

```python
print("Customer SignupDate:", customers["SignupDate"].dtype)
print("Transaction OrderDate:", transactions["OrderDate"].dtype)
```

```python
# saving processed data

customers.to_csv(
    r"D:\Customer Segmentation Analysis\data\processed\customers_cleaned.csv",
    index=False
)

transactions.to_csv(
    r"D:\Customer Segmentation Analysis\data\processed\transactions_cleaned.csv",
    index=False
)

products.to_csv(
    r"D:\Customer Segmentation Analysis\data\processed\products_cleaned.csv",
    index=False
)

annual_summary.to_csv(
    r"D:\Customer Segmentation Analysis\data\processed\annual_summary_cleaned.csv",
    index=False
)

print("All cleaned datasets saved successfully!")
```
